## Questão 1

Considere que temos uma lista de tuplas representando o inventário de um servidor de armazenamento de arquivos. Cada tupla está organizada estritamente na seguinte ordem:
* ID do usuário (str)
* Tipo do arquivo (ex. 'pdf', 'png', 'mp4')
* Tamanho do arquivo em Megabytes (float)
* Dias desde o último acesso (int)

Seja a lista contendo os registros abaixo. Calcule o **tamanho médio** dos arquivos cujo tipo seja exatamente `'pdf'`.

a) 41.25 MB<br>
b) 38.25 MB<br>
c) 35.50 MB<br>
d) 45.00 MB

```python
arquivos = [
    ('usr_01', 'pdf', 45.5, 12),
    ('usr_02', 'png', 4.2, 85),
    ('usr_01', 'mp4', 850.0, 3),
    ('usr_03', 'pdf', 12.0, 120),
    ('usr_04', 'pdf', 65.5, 5),
    ('usr_02', 'pdf', 30.0, 45),
    ('usr_03', 'png', 8.5, 9)
]

totalSizePdf = 0.00;
totalCountPdf = 0;

for arquivo in arquivos:
    if arquivo[1] == 'pdf':
        totalCountPdf = totalCountPdf + 1;
        totalSizePdf = totalSizePdf + arquivo[2]

print(totalSizePdf/totalCountPdf)

#Resposta B - 38.25
```

## Questão 2

Utilizando a mesma lista de tuplas da questão anterior, crie uma nova lista contendo strings formadas pelo ID do usuário e o tipo do arquivo no formato: `ID_tipo` (exemplo: se o ID for `'usr_01'` e o tipo for `'pdf'`, a string resultante será `'usr_01_pdf'`).

A seguir, utilizando conceitos de manipulação de strings ou estruturas de repetição, conte quantas dessas strings geradas possuem **comprimento (tamanho) estritamente maior do que 10**.

a) 1<br>
b) 4<br>
c) 0<br>
d) 2

```python
stringList = []

for arquivo in arquivos:
    stringList.append(arquivo[0] + '_' + arquivo[1])

filteredStringList = [string for string in stringList if len(string) > 10]

print(len(filteredStringList))

# Resposta C - 0
```

## Questão 3

Carregue o arquivo `WorldCupMatches.csv`. Desenvolva uma função que receba o nome de uma seleção e o dataframe de partidas. Essa função deve varrer o arquivo utilizando o método `iterrows()` para calcular **quantos jogos a seleção do 'Brazil' realizou jogando dentro de casa** (coluna `HomeTeamName`) e, dentre esses jogos específicos, **quantos gols ele marcou no total** (coluna `HomeTeamGoals`).

a) 72 jogos; 160 gols,<br>
b) 78 jogos; 176 gols<br>
c) 82 jogos; 180 gols<br>
d) 78 jogos; 171 gols

```python
import pandas as pd
# Certifique-se de ter o arquivo 'WorldCupMatches.csv' na mesma pasta antes de rodar
partidas = pd.read_csv('WorldCupMatches.csv')


def homeTeamScore(teamName, df):
    it = partidas.iterrows()
    count = 0;
    score = 0;
    
    for i, r in it:
        if r['HomeTeamName'] == teamName:
            score = score + r['HomeTeamGoals']
            count = count + 1
            
    return (count, score)

homeTeamScore('Brazil', partidas)

# Resposta B (78, 176)    
```

## Questão 4

Crie uma função lambda que processe uma tupla contendo um registro de funcionário estruturado na forma:
`(nome (str), salario_base (float), departamento (str))`

E que retorne uma nova tupla contendo o nome do funcionário adicionando o sufixo ` - REAJUSTADO` e o salário base com um acréscimo de 10%.

Exemplo de entrada: `('Ana', 5000.0, 'TI')`
Saída da função lambda: `('Ana - REAJUSTADO', 5500.0, 'TI')`

Crie uma função que receba a lista de funcionários e um departamento alvo, e utilize a função lambda dentro de uma *list comprehension* para retornar uma lista contendo **apenas** os funcionários que pertencem ao departamento alvo com seus respectivos salários reajustados.

Considerando a lista `funcionarios` fornecida abaixo e o departamento alvo como `'TI'`, execute a função. Qual é o **salário mínimo e máximo** encontrados na lista resultante de funcionários reajustados?

a) Mínimo: 3300.0; Máximo: 7700.0<br>
b) Mínimo: 3000.0; Máximo: 7000.0<br>
c) Mínimo: 3630.0; Máximo: 8250.0<br>
d) Mínimo: 3630.0; Máximo: 7700.0

```python
funcionarios = [
    ('Ana', 7500.0, 'TI'),
    ('Carlos', 4200.0, 'RH'),
    ('Beatriz', 3300.0, 'TI'),
    ('David', 6000.0, 'Vendas'),
    ('Elena', 5000.0, 'TI')
]

mapping = lambda funcionario: (funcionario[0] + ' - REAJUSTADO', round(funcionario[1] * 1.1, 2), funcionario[2])
reajuste = tuple(map(mapping, funcionarios))

# min_tuple = min(data, key=lambda row: row[1])

minimo = min(reajuste, key=lambda funcionario: funcionario[1])
maximo = max(reajuste, key=lambda funcionario: funcionario[1])

print(minimo, maximo)

# Resposta C - ('Beatriz - REAJUSTADO', 3630.0, 'TI') ('Ana - REAJUSTADO', 8250.0, 'TI')
```
